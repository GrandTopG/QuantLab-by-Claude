```python
import pandas as pd

datos = pd.read_csv("../data/XAUUSD_diario.csv", index_col="Date", parse_dates=True)
datos.head()
```

```python
print(datos.dtypes)
```

```python
print(datos.index.dtype)
```

```python
datos["MA_rapida"] = datos["Close"].rolling(window=20).mean()
datos["MA_lenta"] = datos["Close"].rolling(window=50).mean()

datos[["Close", "MA_rapida", "MA_lenta"]].tail(10)
```

```python
datos["Señal"] = 0
datos.loc[datos["MA_rapida"] > datos["MA_lenta"], "Señal"] = 1
datos.loc[datos["MA_rapida"] < datos["MA_lenta"], "Señal"] = -1

datos[["Close", "MA_rapida", "MA_lenta", "Señal"]].tail(10)
```

```python
datos["Cambio"] = datos["Señal"].diff()

datos[datos["Cambio"] != 0][["Close", "MA_rapida", "MA_lenta", "Señal", "Cambio"]]

```

```python
datos[["Close", "MA_rapida", "MA_lenta", "Señal"]].head(55)
```

```python
datos["Retorno_diario"] = datos["Close"].pct_change()

datos["Retorno_estrategia"] = datos["Señal"].shift(1) * datos["Retorno_diario"]

datos[["Close", "Señal", "Retorno_diario", "Retorno_estrategia"]].tail(10)
```

```python
datos["Retorno_acumulado"] = (1 + datos["Retorno_estrategia"]).cumprod()

datos[["Close", "Retorno_estrategia", "Retorno_acumulado"]].tail(10)
```

```python
datos["Tramo"] = (datos["Cambio"] != 0).cumsum()
```

```python
tabla = datos.reset_index()

resumen_tramos = tabla.groupby("Tramo").agg(
    Fecha_inicio=("Date", "min"),
    Fecha_fin=("Date", "max"),
    Dias=("Date", "count"),
    Señal=("Señal", "first"),
    Retorno_total=("Retorno_estrategia", lambda x: (1 + x).prod() - 1)
)

resumen_tramos
```

```python
import plotly.graph_objects as go

fig = go.Figure()
fig.add_trace(go.Scatter(x=datos.index, y=datos["Retorno_acumulado"], mode="lines", name="Estrategia"))

fig.update_layout(title="Curva de Capital - Estrategia Cruce de Medias (XAUUSD)",
                   xaxis_title="Fecha", yaxis_title="Capital (base $1)")
fig.show()
```

```python
datos["Maximo_historico"] = datos["Retorno_acumulado"].cummax()
datos["Drawdown"] = (datos["Retorno_acumulado"] / datos["Maximo_historico"]) - 1

drawdown_maximo = datos["Drawdown"].min()
print(f"Drawdown máximo: {drawdown_maximo:.2%}")
```

```python
operaciones = resumen_tramos[resumen_tramos["Señal"] != 0]

ganadoras = operaciones[operaciones["Retorno_total"] > 0]

win_rate = len(ganadoras) / len(operaciones)
print(f"Total de operaciones: {len(operaciones)}")
print(f"Operaciones ganadoras: {len(ganadoras)}")
print(f"Win rate: {win_rate:.2%}")
```

```python
retorno_promedio_diario = datos["Retorno_estrategia"].mean()
volatilidad_diaria = datos["Retorno_estrategia"].std()

sharpe_ratio = (retorno_promedio_diario / volatilidad_diaria) * (252 ** 0.5)

print(f"Retorno promedio diario: {retorno_promedio_diario:.4%}")
print(f"Volatilidad diaria: {volatilidad_diaria:.4%}")
print(f"Sharpe Ratio (anualizado): {sharpe_ratio:.2f}")
```
