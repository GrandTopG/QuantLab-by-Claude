```python
import yfinance as yf

datos = yf.download("GC=F", period="1y", interval="1d")
datos.head()
```

```python
print(f"Total de días descargados: {len(datos)}")
print(f"Desde: {datos.index.min()} hasta: {datos.index.max()}")
```

```python
print(datos.columns)
```

```python
datos = datos[["Open", "High", "Low", "Close", "Volume"]]
datos.head()
```

```python
datos.to_csv("../data/XAUUSD_diario.csv")
print("Datos guardados correctamente")
```

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Candlestick(
    x=datos.index,
    open=datos["Open"],
    high=datos["High"],
    low=datos["Low"],
    close=datos["Close"]
)])

fig.update_layout(title="XAUUSD - Velas Diarias (1 año)", xaxis_title="Fecha", yaxis_title="Precio (USD)")
fig.show()
```
