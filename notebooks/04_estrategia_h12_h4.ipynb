```python
import yfinance as yf

datos_h1 = yf.download("GC=F", period="730d", interval="1h")
datos_h1.head()
```

```python
datos_h1.columns = datos_h1.columns.get_level_values(0)
datos_h1.columns.name = None
datos_h1 = datos_h1[["Open", "High", "Low", "Close", "Volume"]]

datos_h1.head()
```

```python
datos_h1.to_csv("../data/XAUUSD_h1.csv")
print("Datos H1 guardados correctamente")
```

```python
datos_h4 = datos_h1.resample("4h", offset="17h").agg({
    "Open": "first",
    "High": "max",
    "Low": "min",
    "Close": "last",
    "Volume": "sum"
})

datos_h4 = datos_h4.dropna()

datos_h4.head()
```

```python
datos_h12 = datos_h1.resample("12h", offset="17h").agg({
    "Open": "first",
    "High": "max",
    "Low": "min",
    "Close": "last",
    "Volume": "sum"
})

datos_h12 = datos_h12.dropna()

datos_h12.head()
```

```python
datos_h4.to_csv("../data/XAUUSD_h4.csv")
datos_h12.to_csv("../data/XAUUSD_h12.csv")
print("Velas H4 y H12 guardadas correctamente")
```

```python
datos_h12["C1"] = datos_h12["Close"]
datos_h12["C2"] = datos_h12["Close"].shift(1)
datos_h12["C3"] = datos_h12["Close"].shift(2)

datos_h12[["Close", "C1", "C2", "C3"]].head(10)
```

```python
datos_h12["Es_roja"] = datos_h12["Close"] < datos_h12["Open"]

datos_h12["OURAS"] = datos_h12["Open"].where(datos_h12["Es_roja"]).ffill()

datos_h12[["Open", "Close", "Es_roja", "OURAS"]].head(15)
```

```python
datos_h12.to_csv("../data/XAUUSD_h12_completo.csv")
print("Datos H12 con C1, C2, C3 y OURAS guardados")
```

```python
datos_h12["Escenario_A"] = (datos_h12["C1"] > datos_h12["C2"]) & (datos_h12["C1"] > datos_h12["C3"])
datos_h12["Escenario_B"] = (datos_h12["C1"] < datos_h12["C2"]) & (datos_h12["C1"] > datos_h12["C3"])

datos_h12[["C1", "C2", "C3", "Escenario_A", "Escenario_B"]].tail(10)
```

```python
datos_h12["Sobre_OURAS_C1"] = datos_h12["C1"] > datos_h12["OURAS"]
datos_h12["Sobre_OURAS_C2"] = datos_h12["C2"] > datos_h12["OURAS"]
datos_h12["Sobre_OURAS_C3"] = datos_h12["C3"] > datos_h12["OURAS"]

datos_h12["Conteo_sobre_OURAS"] = datos_h12["Sobre_OURAS_C1"].astype(int) + datos_h12["Sobre_OURAS_C2"].astype(int) + datos_h12["Sobre_OURAS_C3"].astype(int)

datos_h12["Sub_caso"] = "N/A"
datos_h12.loc[datos_h12["Conteo_sobre_OURAS"] == 3, "Sub_caso"] = "1_Limpio"
datos_h12.loc[datos_h12["Conteo_sobre_OURAS"] == 0, "Sub_caso"] = "2_Sin_cruzar"
datos_h12.loc[(datos_h12["Conteo_sobre_OURAS"] == 1) | (datos_h12["Conteo_sobre_OURAS"] == 2), "Sub_caso"] = "3_Cruzando"

datos_h12[["C1", "C2", "C3", "OURAS", "Conteo_sobre_OURAS", "Sub_caso"]].tail(10)
```

```python
datos_h12.to_csv("../data/XAUUSD_h12_completo.csv")
print("Escenarios y Sub-casos guardados")
```

```python
datos_h12["Combinacion"] = "Ninguno"
datos_h12.loc[datos_h12["Escenario_A"], "Combinacion"] = "A_" + datos_h12["Sub_caso"]
datos_h12.loc[datos_h12["Escenario_B"], "Combinacion"] = "B_" + datos_h12["Sub_caso"]

datos_h12["Retorno_vela_siguiente"] = datos_h12["Close"].pct_change().shift(-1)
datos_h12["Siguiente_fue_alcista"] = datos_h12["Close"].shift(-1) > datos_h12["Open"].shift(-1)

resumen_escenarios = datos_h12[datos_h12["Combinacion"] != "Ninguno"].groupby("Combinacion").agg(
    Total_casos=("Combinacion", "count"),
    Pct_siguiente_alcista=("Siguiente_fue_alcista", "mean"),
    Retorno_promedio_siguiente=("Retorno_vela_siguiente", "mean")
)

resumen_escenarios
```

```python
from scipy import stats

combinaciones = ["A_1_Limpio", "A_2_Sin_cruzar", "A_3_Cruzando", "B_2_Sin_cruzar", "B_3_Cruzando"]

for combo in combinaciones:
    retornos_combo = datos_h12[datos_h12["Combinacion"] == combo]["Retorno_vela_siguiente"].dropna()
    t_stat, p_value = stats.ttest_1samp(retornos_combo, 0)
    print(f"{combo}: n={len(retornos_combo)}, t={t_stat:.3f}, p-value={p_value:.4f}")
```

```python
datos_h12["Año"] = datos_h12.index.year

primer_periodo = datos_h12[datos_h12["Año"] == datos_h12["Año"].min()]
segundo_periodo = datos_h12[datos_h12["Año"] > datos_h12["Año"].min()]

for nombre, periodo in [("Primer año", primer_periodo), ("Resto del periodo", segundo_periodo)]:
    casos_a3 = periodo[periodo["Combinacion"] == "A_3_Cruzando"]["Retorno_vela_siguiente"].dropna()
    if len(casos_a3) > 1:
        t_stat, p_value = stats.ttest_1samp(casos_a3, 0)
        print(f"{nombre}: n={len(casos_a3)}, retorno promedio={casos_a3.mean():.4%}, p-value={p_value:.4f}")
```

```python
datos_h12["Año"] = datos_h12.index.year

combinaciones = ["A_1_Limpio", "A_2_Sin_cruzar", "A_3_Cruzando", "B_2_Sin_cruzar", "B_3_Cruzando"]

for combo in combinaciones:
    print(f"\n--- {combo} ---")
    for año in sorted(datos_h12["Año"].unique()):
        subset = datos_h12[(datos_h12["Combinacion"] == combo) & (datos_h12["Año"] == año)]
        retornos_subset = subset["Retorno_vela_siguiente"].dropna()
        if len(retornos_subset) > 1:
            t_stat, p_value = stats.ttest_1samp(retornos_subset, 0)
            print(f"  Año {año}: n={len(retornos_subset)}, retorno promedio={retornos_subset.mean():.4%}, p-value={p_value:.4f}")
        else:
            print(f"  Año {año}: muestra insuficiente (n={len(retornos_subset)})")
```
