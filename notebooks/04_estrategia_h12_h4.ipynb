```python
import yfinance as yf

datos_h1 = yf.download("GC=F", period="730d", interval="1h")
datos_h1.head()
```

```python
datos_h1.columns = datos_h1.columns.get_level_values(0)
datos_h1.columns.name = None
datos_h1 = datos_h1[["Open", "High", "Low", "Close", "Volume"]]

datos_h1.head()
```

```python
datos_h1.to_csv("../data/XAUUSD_h1.csv")
print("Datos H1 guardados correctamente")
```
